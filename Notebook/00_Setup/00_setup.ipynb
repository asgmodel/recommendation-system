## Repository Setup
Use the existing repository and its configured GitHub credentials.

```python
from pathlib import Path
import subprocess
import sys

REPOSITORY = "https://github.com/asgmodel/recommendation-system.git"
candidates = (Path.cwd(), *Path.cwd().parents)
ROOT = next((p for p in candidates if (p / "src/hlo_recommender").is_dir()), None)
if ROOT is None:
    ROOT = Path.cwd() / "recommendation-system"
    if not ROOT.exists():
        subprocess.run(["git", "clone", REPOSITORY, str(ROOT)], check=True)
    if not (ROOT / "src/hlo_recommender").is_dir():
        raise FileNotFoundError("Select the recommendation-system checkout.")
ROOT.name
```
